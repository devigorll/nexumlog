# Código:

### Importando bibliotecas


```python
import os
from dotenv import load_dotenv
from supabase import create_client, Client
import pandas as pd
import numpy as np
```

### Conectando ao banco

```python

# Força o recarregamento das variáveis do .env
load_dotenv(override=True)

url = os.getenv("SUPABASE_URL")
key = os.getenv("SUPABASE_KEY")

supabase: Client = create_client(url, key)

```

### Inserindo status de pedido

```python
status = [

    {
        "nome_status": "Aguardando liberacao",
        "descricao": "Pedido criado, aguardando documentacao ou veiculo"
    },
    {
        "nome_status": "Em carregamento",
        "descricao": "Carga sendo alocada no caminhao"
    },
    {
        "nome_status": "Em transporte",
        "descricao": "Veiculo em transito para o porto/destino"
    },
    {
        "nome_status": "Entregue",
        "descricao": "Carga entregue no destino final"
    },
    {
        "nome_status": "Atrasado",
        "descricao": "Alerta de atraso na logistica"
    },
    {
        "nome_status": "Cancelado",
        "descricao": "Pedido cancelado"
    }

]

resposta = supabase.table("status_pedido_tb").insert(status).execute()
print("Valores inseridos com sucesso:", resposta.data)
```

```python
resposta = supabase.table("status_pedido_tb").select("*").execute()

df = pd.DataFrame(resposta.data)
df
```
